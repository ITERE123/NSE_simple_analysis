```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pd.read_csv(r"C:\Users\oweni\Downloads\archive (1)\Teen_Mental_Health_Dataset.csv")
data.head()
```

px.

```python
data_clean = data.dropna(subset=["gender", "age"])
fig = px.box(
data_clean,
x="gender",
y="age",
title="Age distribution by gender",
labels={"gender": "Gender", "age": "Age"}
)
fig.show()
```

```python
sns.boxplot(x="gender", y="age", data=data)
plt.xlabel("gender")
plt.ylabel("age")
plt.title("Age distribution by gender")
plt.show()
```

```python
sns.violinplot(x="gender",y="daily_social_media_hours",data=data)
plt.title("Daily social media spent by each gender")
plt.show()
```

```python
data_clean = data.dropna(subset=["gender", "daily_social_media_hours"])
fig = px.box(
data_clean,
x="gender",
y="daily_social_media_hours",
title="Age distribution by gender",
labels={"gender": "Gender", "age": "Age"}
)
fig.show()
```

```python

fig = px.histogram(data, x="platform_usage", nbins=100, title="Distribution of platform usage",marginal="box", labels={"platform_usage": "Platform Usage (hours)"})
fig.show()
```

```python

sns.scatterplot(x="daily_social_media_hours",y="sleep_hours",data=data)
plt.title("Relationship between daily social media hours and sleep hours")
plt.xlabel("Daily Social Media Hours")
plt.ylabel("Sleep Hours")
plt.show()
```

```python
px.scatter(data, x="screen_time_before_sleep", y="sleep_hours", title="Relationship between screen time before sleep and sleep hours", labels={"screen_time_before_sleep": "Screen Time Before Sleep", "sleep_hours": "Sleep Hours"}).show()
```

```python
fig = px.scatter(data, 
                 x="screen_time_before_sleep", 
                 y="sleep_hours",
                 trendline="ols",  # this draws the regression line for you
                 title="Does Screentime Before Sleep Affect Sleep Hours?")
fig.show()
```

```python
fig = px.scatter(data, x="daily_social_media_hours", y="academic_performance", title="Relationship between daily social media hours and academic performance", labels={"daily_social_media_hours": "Daily Social Media Hours", "academic_performance": "Academic Performance"}, trendline="ols")
fig.show()
```

```python
fig = px.box(
    data,
    y="daily_social_media_hours",
    title="Distribution of Daily Social Media Hours",
    labels={"daily_social_media_hours": "Daily Social Media Hours"}
)
fig.show()

```
